from heapq import heappush, heappop
from bisect import bisect_left

class Solution:
    """
    Problem: Isomorphic Strings (LeetCode 205)
    
    Two strings are isomorphic if characters in s can be replaced to get t,
    maintaining a one-to-one mapping (bijection).
    """
    
    # ═══════════════════════════════════════════════════════════════
    # APPROACH 1: Two Hash Maps (Most Intuitive)
    # ═══════════════════════════════════════════════════════════════
    
    def isIsomorphic_two_maps(self, s: str, t: str) -> bool:
        """
        Time Complexity: O(n) - single pass through strings
        Space Complexity: O(n) - two dictionaries storing mappings
        
        Mental Model: Maintain bidirectional mapping
        - s_to_t: maps characters from s to t
        - t_to_s: maps characters from t to s (reverse mapping)
        
        Why two maps?
        - s_to_t ensures each char in s maps to one char in t
        - t_to_s ensures no two chars in s map to same char in t
        """
        if len(s) != len(t):
            return False
        
        s_to_t = {}  # Forward mapping: s[i] -> t[i]
        t_to_s = {}  # Reverse mapping: t[i] -> s[i]
        
        # Use enumerate - proper Pythonic iteration with index
        for i, (char_s, char_t) in enumerate(zip(s, t)):
            
            # Check forward mapping
            if char_s in s_to_t:
                if s_to_t[char_s] != char_t:
                    return False  # Inconsistent mapping
            else:
                s_to_t[char_s] = char_t
            
            # Check reverse mapping (prevents two-to-one)
            if char_t in t_to_s:
                if t_to_s[char_t] != char_s:
                    return False  # Two different chars map to same target
            else:
                t_to_s[char_t] = char_s
        
        return True
    
    
    # ═══════════════════════════════════════════════════════════════
    # APPROACH 2: Single Map + Value Check (Your Attempted Approach)
    # ═══════════════════════════════════════════════════════════════
    
    def isIsomorphic_single_map(self, s: str, t: str) -> bool:
        """
        Time Complexity: O(n²) - checking if value exists is O(n) per iteration
        Space Complexity: O(n)
        
        This is what you were trying to do, but correctly implemented.
        Note: Less efficient than two-map approach due to value searches.
        """
        if len(s) != len(t):
            return False
        
        mapping = {}
        
        for i in range(len(s)):  # Proper index-based iteration
            char_s, char_t = s[i], t[i]
            
            if char_s in mapping:
                # Character already mapped - check consistency
                if mapping[char_s] != char_t:
                    return False
            else:
                # New character - check if target already used
                if char_t in mapping.values():  # O(n) check
                    return False  # Another char already maps to this target
                mapping[char_s] = char_t
        
        return True
    
    
    # ═══════════════════════════════════════════════════════════════
    # APPROACH 3: Transformation Pattern (Most Elegant)
    # ═══════════════════════════════════════════════════════════════
    
    def isIsomorphic_pattern(self, s: str, t: str) -> bool:
        """
        Time Complexity: O(n)
        Space Complexity: O(n)
        
        Genius Insight: Two strings are isomorphic if they have the
        same "transformation pattern".
        
        Pattern Concept:
        - "paper" → [0, 1, 0, 2, 3] (positions of first occurrence)
        - "title" → [0, 1, 0, 2, 3] (same pattern!)
        
        How it works:
        - For each character, record its first occurrence index
        - If patterns match, strings are isomorphic
        """
        def get_pattern(string):
            """
            Maps each character to the index of its first occurrence.
            Example: "egg" → [0, 1, 1] (e at 0, g first seen at 1)
            """
            seen = {}
            pattern = []
            for i, char in enumerate(string):
                if char not in seen:
                    seen[char] = i
                pattern.append(seen[char])
            return pattern
        
        return get_pattern(s) == get_pattern(t)
    
    
    # ═══════════════════════════════════════════════════════════════
    # APPROACH 4: Index Mapping (Space Optimized)
    # ═══════════════════════════════════════════════════════════════
    
    def isIsomorphic_optimal(self, s: str, t: str) -> bool:
        """
        Time Complexity: O(n)
        Space Complexity: O(1) - fixed size arrays (256 for ASCII)
        
        Expert Technique: Use arrays instead of hash maps for small
        character sets.
        
        Arrays store the last seen index of each character.
        If indices don't match for corresponding positions, not isomorphic.
        """
        if len(s) != len(t):
            return False
        
        # Arrays to store last seen index (256 for extended ASCII)
        s_last_seen = [-1] * 256
        t_last_seen = [-1] * 256
        
        for i in range(len(s)):
            char_s_code = ord(s[i])
            char_t_code = ord(t[i])
            
            # Check if both characters have the same "last seen" index
            if s_last_seen[char_s_code] != t_last_seen[char_t_code]:
                return False
            
            # Update last seen indices
            s_last_seen[char_s_code] = i
            t_last_seen[char_t_code] = i
        
        return True

## 📚 Problem Understanding: Isomorphic Strings

**Definition**: Two strings are **isomorphic** if characters in `s` can be replaced to get `t`, maintaining a **one-to-one mapping** (bijection).

**Key Constraints:**
1. Each character in `s` maps to **exactly one** character in `t` (function)
2. **No two different** characters in `s` can map to the **same** character in `t` (injection)

**Visual Example:**
```
"paper" → "title"
p → t
a → i
p → t (✓ consistent)
e → l
r → e

This is VALID because:
- Each char in s has one mapping
- No two different chars in s share the same target
```

```
"badc" → "baba"
b → b
a → a
d → b (✗ CONFLICT! 'b' and 'd' both try to map to 'b')
c → a (✗ CONFLICT! 'a' and 'c' both try to map to 'a')

This is INVALID
```

## 🔄 Rust Implementation (Performance-Focused)

```rust
use std::collections::HashMap;

impl Solution {
    /// Two HashMap approach - most straightforward
    pub fn is_isomorphic_two_maps(s: String, t: String) -> bool {
        if s.len() != t.len() {
            return false;
        }
        
        let mut s_to_t: HashMap<char, char> = HashMap::new();
        let mut t_to_s: HashMap<char, char> = HashMap::new();
        
        // Zip iterators for parallel iteration
        for (char_s, char_t) in s.chars().zip(t.chars()) {
            // Check s -> t mapping
            if let Some(&mapped) = s_to_t.get(&char_s) {
                if mapped != char_t {
                    return false;
                }
            } else {
                s_to_t.insert(char_s, char_t);
            }
            
            // Check t -> s mapping (prevents many-to-one)
            if let Some(&mapped) = t_to_s.get(&char_t) {
                if mapped != char_s {
                    return false;
                }
            } else {
                t_to_s.insert(char_t, char_s);
            }
        }
        
        true
    }
    
    /// Array-based approach - O(1) space for ASCII
    pub fn is_isomorphic_optimal(s: String, t: String) -> bool {
        if s.len() != t.len() {
            return false;
        }
        
        let mut s_last_seen = [-1i32; 256];
        let mut t_last_seen = [-1i32; 256];
        
        for (i, (char_s, char_t)) in s.chars().zip(t.chars()).enumerate() {
            let s_idx = char_s as usize;
            let t_idx = char_t as usize;
            
            if s_last_seen[s_idx] != t_last_seen[t_idx] {
                return false;
            }
            
            s_last_seen[s_idx] = i as i32;
            t_last_seen[t_idx] = i as i32;
        }
        
        true
    }
}
```
